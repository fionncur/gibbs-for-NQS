# **Exercise 2: Gibbs sampling of a restricted Boltzmann machine**

Consider visible spins $\mathbf{v} \in \{−1,+1\}^{N_v}$ and hidden spins $\mathbf{h} \in \{−1,+1\}^{N_h}$, with energy
$$
H(\mathbf{v},\mathbf{h}) = -\sum_i a_i v_i + -\sum_j b_j h_j + -\sum_{i,\, j} W_{ij} v_i h_j
$$
The sums over $i,j$ occur over the components of the tuples in the relevant domains (over plus and minus one, of length $N_v, N_h$).
Their joint distribution is
$$
\pi_\beta (\mathbf{v} , \mathbf{h}) = \frac{1}{Z(\beta)} e^{-\beta H(\mathbf{v},\mathbf{h})}
$$
There are interactions between layers but no interactions within either layer.


Its parameters are trained so that the *visible marginal distribution* matches the observed distribution. The former is a sum over all configurations of $\mathbf{h}$ in the joint distribution
$$
\pi_\beta(\mathbf{v}) = \sum_\mathbf{h} \pi_{\beta}(\mathbf{v},\mathbf{h}).
$$

**(a)** Explain why the absence of interactions within each RBM layer gives
$$
\pi(\mathbf{h}|\mathbf{v})= \Pi_{j} \pi (h_j|\mathbf{v}) \qquad \pi(\mathbf{v}|\mathbf{h})= \Pi_{i} \pi (v_i|\mathbf{h})
$$
Are the spins within one layer also independent in the marginal distribution?

 ----

 - Here, the "absence of interactions" within a layer means that the energy function $H$ of the entire RBM state (i.e. $\mathbf{h,v}$) behaves as such: fixing a state $\mathbf{h,v}$, and independently varying a *single component* or node, (in our domain this means flipping a sign e.g. $v_i \rightarrow -v_i$ and likewise for a hidden node), results in an energy difference $\Delta H$ that is entirely independent of the fixed state of the components *within its own layer*, i.e. flipping an $si$ produces a $\Delta H$ depending only on the weights and $\mathbf{h}$, and vice-versa. 
 
 - However this may obviously depend on the state (of index $i$ or $j$) we choose to vary, by virtue of the correspondng coefficient/weight. This would not be the case if we had terms in the function $H$ proportional to, say,  $v_iv_j$ for distinct $i,j$. 

 - why the conditional distributions factorize as given can be easily derived from the form of the RBM:

$$
\begin{align*}
\pi(\mathbf{h}|\mathbf{v})  &= \frac{\pi(\mathbf{v},\mathbf{h})}{\pi(\mathbf{v})} \\
                            &= \frac{\pi(\mathbf{v},\mathbf{h})}{\sum_{\mathbf{h'}}\pi(\mathbf{v,h'})}
\end{align*}
$$
where the sum $\mathbf{h'}$ occurs in the domain $\{-1,+1\}^{N_h}$. Sub in the form for $\pi$ and note that the factors $Z$ cancel
$$
\begin{align*}
\pi(\mathbf{h}|\mathbf{v})  &= \frac{e^{-\beta H (\mathbf{v,h})}}{\sum_{\mathbf{h'}} e^{-\beta H (\mathbf{v,h'})}} \\
                            &= \frac{e^{-\beta(\sum_i a_i v_i + \sum_j b_j h_j + \sum_{i,\, j} W_{ij} v_i h_j)}}{\sum_{\mathbf{h'}} e^{-\beta(\sum_i a_i v_i + \sum_j b_j h'_j + \sum_{i,\, j} W_{ij} v_i h'_j)}} \\
                            &\text{factor out the $a_iv_i$ terms and divide} \\
                            &= \frac{e^{-\beta(\sum_j b_j h_j + \sum_{i,\, j} W_{ij} v_i h_j)}}{\sum_{\mathbf{h'}} e^{-\beta(\sum_j b_j h'_j + \sum_{i,\, j} W_{ij} v_i h'_j)}} \\
                            &= \frac{e^{-\beta(\sum_j h_j(b_j + \sum_{i} W_{ij} v_i)}}{\sum_{\mathbf{h'}} e^{-\beta(\sum_j h'_j(b_j + \sum_{i} W_{ij} v_i))}} 
\end{align*}
$$
the numerator factorizes simply 
$$
e^{-\beta( \sum_j h_j(b_j + \sum_{i} W_{ij} v_i) } = \prod_j e^{\beta(h_j(b_j + \sum_i W_{ij} v_i ))}.
$$
For the denominator,
$$
\sum_{\mathbf{h'}} e^{-\beta(\sum_j h'_j(b_j + \sum_{i} W_{ij} v_i))} = \sum_{h'_1 = \pm 1} \cdots \sum_{h'_{N_h} = \pm 1} \prod_j e^{h'_j (b_j + \sum_i W_{ij} v_i)}.
$$
Since each term in the product features only one $j$, we can collapse the sums into a single one and exchange the order
$$
\sum_{\mathbf{h'}} e^{-\beta(\sum_j h'_j(b_j + \sum_{i} W_{ij} v_i))} = \prod_j \left(\sum_{h'_j = \pm 1} e^{-\beta(h'_j (b_j + \sum_i W_{ij} v_i))}\right).
$$
now 
$$
\pi_\beta (\mathbf{h}|\mathbf{v}) = \prod_j \frac{e^{-\beta(h_j(b_j + \sum_i W_{ij} v_i ))}}{\sum_{h'_j = \pm 1} e^{-\beta(h'_j (b_j + \sum_i W_{ij} v_i))}}
$$
The term inside the product is precisely the single state conditional probability
$$
\pi_\beta (\mathbf{h}|\mathbf{v}) = \prod_j P(h_j | \mathbf{v}).
$$
the argument for $\pi(\mathbf{v} | \mathbf{h})$ is identical.

**(b)** just trivial algebra from the form of the single state conditional probability

**Exercise 2.2. Vectorized implementation** Consult scikit-learn’s ``BernoulliRBM.gibbs`` documentation.
Click ``[source]`` beside the method to open its implementation on GitHub, and inspect ``gibbs``,
``_sample_hiddens``, ``_mean_hiddens``, and ``_sample_visibles``. The implementation combines matrix operations with arrays of uniform random numbers.

Implement the same vectorised sampling structure in your chosen language. Scikit-learn uses
binary variables in $\{0,1\}$; this exercise uses spins in $\{−1,+1\}$. Use the conditional probabilities
derived above.
For $B$ independently initialised chains, let $V$ have shape $B \times n_v$, let $R$ store hidden spins with
shape $B\times n_h$, and let $W$ have shape $n_v \times n_h$.

Repeat for the requested number of sweeps:

 -----
Compute $P_h = \sigma(2\beta[VW + \text{hidden biases}])$.

Draw independent uniform random numbers $U_h$.

Set each hidden spin to $+1$ if $U_h < P_h$, and $−1$ otherwise.

Compute $P_v = \sigma(2\beta[RW^T + \text{visible biases}])$.

Draw independent uniform random numbers $U_v$.

Set each visible spin to $+1$ if $U_v < P_v$ , and $−1$ otherwise.

Record observables from the current visible and hidden spins.

Biases are broadcast across rows, and the sigmoid acts elementwise.

 -----

The vectorization is exploiting (i) the statistical independence of the spins $\mathbf{h}$ conditional on $\mathbf{v}$, and vice-versa, and also (ii) the independence of the $B$ separate chains.

The formulae for part (b) were 

$$
\begin{align}
    &P(h_j = +1\, |\, \mathbf{v}) = \sigma\left[2\beta\left( b_j + \sum_iW_{ij}v_i\right) \right], \\
    &P(v_i = +1\, |\, \mathbf{h}) = \sigma\left[2\beta\left( a_i + \sum_jW_{ij}h_j\right) \right], \qquad \text{where } \sigma(z) = \frac{1}{1+e^{-z}} 
\end{align}
$$

so to clarify what the pseudocode is doing, define $\vec{a} = [a_1,...,a_{n_v}],\, \vec{b} = [b_1, ..., b_{n_h}]$. 

The matrix $V$ is holding, in each of its $B$ rows, a state of $n_v$ visible spins, and similarly $R$ is holding, in each of its $B$ rows, a state of $n_h$ hidden spins.

Then the matrix $P_h$ is computing exactly expression (1), except in parallel for the $B$ separate chains. Similarly, $P_v$ is computing (2). The "hidden/visible biases" referred to are actually a matrix, with $B$ rows, which are just $B$ copies of the row vector, either $\vec{a}$ or $\vec{b}$.
